```sql
%sql
CREATE OR REPLACE FUNCTION f1_dev.gold.mask_mclaren(name STRING)
RETURNS STRING
RETURN CASE 
  WHEN name = 'Lewis Hamilton' THEN '***'
  ELSE name
END;
```

```sql
ALTER TABLE f1_dev.gold.driver_wins ALTER COLUMN name SET MASK f1_dev.gold.mask_mclaren;
```
