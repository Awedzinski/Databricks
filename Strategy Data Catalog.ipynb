```sql


-- Select Namespace to connect to 
  select distinct concat('`',table_catalog,'`','.','`',TABLE_schema,'`') as NAME_SPACE  from system.information_schema.tables 
  where 
  --table_catalog = 'demo'  and 
  table_schema = 'gold' order by 1;

```

```sql
-- list tables and columns from the workspace above

 select concat('`',table_catalog,'`','.','`',TABLE_schema,'`','.') as NAME_SPACE, concat('`',TABLE_NAME,'`') as TAB_NAME, column_name COL_NAME, data_type DATA_TYPE, CASE WHEN data_type IN ('TINYINT','BYTE') THEN 1 WHEN data_type IN ('BIGINT', 'LONG','DOUBLE') THEN 8 WHEN data_type IN ('INT','FLOAT') THEN 4 WHEN data_type IN ('SMALLINT','SHORT') THEN 2 WHEN data_type IN ('DATE') THEN 10 WHEN data_type IN ('TIMESTAMP') THEN 26 ELSE -1 END AS DATA_LEN
 from system.information_schema.columns 
 where 
 --table_catalog = 'demo' and 
 table_schema = 'gold'
  order by 1,2
```
